```python
def mergesort(arr1):
  if len(arr1)==1:
    return arr1
  mid=len(arr1)//2

  left = mergesort(arr1[:mid])             # Merge sort
  right= mergesort(arr1[mid:])

  return merge(left,right)

def merge(left,right):
  i=0
  j=0
  ans=[]
  while(i<len(left) and j< len(right)):
    if left[i]<right[j]:
      ans.append(left[i])
      i+=1
    else:
      ans.append(right[j])
      j+=1
  while(i<len(left)):
    ans.append(left[i])
    i+=1
  while(j<len(right)):
      ans.append(right[j])
      j+=1
  return ans
if __name__=="__main__":
  arr1=[5,1,8,9,3,2,6]
  print(mergesort(arr1))

```

```python
def solve(x,y,maze,n,ans,path,visited):

  if x==n-1 and y==n-1 :
    ans.append(path)
    print(ans)
    return

  #Down
  if(x+1<n and visited[x][y]!=1 and maze[x+1][y]==1):
    visited[x][y]==1
    solve(x+1,y,maze,n,ans,path+'D',visited)
    visited[x][y]==0


  #Left
  if(y-1>=0 and visited[x][y-1]!=1 and maze[x][y-1]==1):
    visited[x][y]==1
    solve(x+1,y,maze,n,ans,path+'L',visited)
    visited[x][y]==0

  #Right
  if(y+1>=0 and visited[x][y+1]!=1 and maze[x][y+1]==1):
    visited[x][y]==1
    solve(x+1,y,maze,n,ans,path+'R',visited)
    visited[x][y]==0

  #Up
  if(x-1>=0 and visited[x-1][y]!=1 and maze[x-1][y]==1):
    visited[x][y]==1
    solve(x+1,y,maze,n,ans,path+'U',visited)
    visited[x][y]==0


if __name__=="__main__":
  maze=[[1,0,0,0],[1,1,0,1],[1,1,0,0],[0,1,1,1]]
  ans=[]
  x=0
  y=0
  n=len(maze)
  visited= maze
  for i in range(n):
    for j in range(n):
      visited[i][j]=0
  solve(x,y,maze,n,ans,'',visited)











```

Given an integer array nums that may contain duplicates, return all possible subsets (the power set).

The solution set must not contain duplicate subsets. Return the solution in any order.



Example 1:

Input: nums = [1,2,2]
Output: [[],[1],[1,2],[1,2,2],[2],[2,2]]

```python
class Solution(object):
    def subsetsWithDup(self, nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        res=[]
        helper=[]
        index=0
        nums.sort()
        self.subset(index,nums,res,helper)
        return res
    def subset(self,index,nums,res,helper):
        if index==len(nums):
            res.append(helper[:])
            return
        helper.append(nums[index])
        self.subset(index+1,nums,res,helper)
        helper.pop()

        while(index+1 <len(nums) and nums[index]==nums[index+1]):
            index=index+1
        self.subset(index+1,nums,res,helper)
```
